```python
import pandas as pd 
import numpy as np 
import json
```

```python
# Load the subscriptions data
df_subs = pd.read_csv('../data/raw_subscriptions.csv')
```

```python
# Look at the first 10 rows
print("--- FIRST 10 ROWS ---")
df_subs.head(10)
```

```python
# Look at information about data types
print("\n--- DATA INFO ---")
df_subs.info()
```

```python
# Check what plan types we have
print(df_subs['plan_type'].unique())
```

```python
# Noticed that the plan type "individual_typo" looks like a typo. Let's check which rows have this value.
df_subs.query('plan_type == "indivual_typo"')
```

```python
# Fix the plan_type typos using a mapping dictionary
plan_mapping = {
    'Individual': 'Individual',
    'Family': 'Family',
    'Student': 'Student',
    'indivual_typo': 'Individual'
}
df_subs['plan_type'] = df_subs['plan_type'].map(plan_mapping)
```

```python
# Noticed from the Data Type Info that the signup_date and cancellation_date columns are being read in as objects (strings) instead of datetime. Let's convert those.
df_subs['signup_date'] = pd.to_datetime(df_subs['signup_date'])
df_subs['cancellation_date'] = pd.to_datetime(df_subs['cancellation_date'])
```

```python
# Let's verify that our transformations for plan_type worked
print(df_subs['plan_type'].unique())
```

```python
# Let's verify that our traansformation for the date columns worked by looking at the data types again
df_subs.info()
```

```python
# Load activity logs from JSON file
df_logs = pd.read_json('../data/raw_activity_logs.json')
```

```python
# Read first 10 rows of activity logs
print("\n--- FIRST 10 ROWS ---")
df_logs.head(10)
```

```python
# Check Data types of activity logs
print("\n---DATA INFO---")
df_logs.info()
```

```python
# Check device types in activity logs
df_logs['device_type'].unique()
```

```python
# Convert activity_date from string to datetime
df_logs['activity_date'] = pd.to_datetime(df_logs['activity_date'])
```

```python
# Verify change by checking data types again
print("\n--- UPDATED DATA TYPES ---")
df_logs.info()
```

```python
# Check number of rows in activity logs df before merge
df_logs.shape
```

```python
# Like a SQL join, we will merge the subscriptions and activity logs dataframes together using the user_id column
df_master = pd.merge(df_logs, df_subs, on='user_id', how='left')
```

```python
# Let's look at the master dataframe to verify the merge
print('\n--- MASTER DATAFRAME ---')
df_master
```

```python
# Check number of rows in master df after merge
df_master.shape
```
